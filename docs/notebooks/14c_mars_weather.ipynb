### Background: what is going on with Mars's air?

**1. A barometer on Mars reads less every year.** Curiosity is a car-sized rover. It landed in
Gale crater in August 2012, and its weather station has read the air pressure every day since.
In its first Mars year (from 2012) its highest reading was 925 Pa; in its
seventh (to 2024) the highest was 871 Pa, 6% less. Taken at
face value, Mars is losing its air fast. This project finds out whether it is.

<img src="https://raw.githubusercontent.com/AI4EPS/EPS88_PyEarth/main/data/images/T13/curiosity_selfie_mount_sharp.jpg" width="720">

*Curiosity on the lower slopes of Mount Sharp in January 2015: a self-portrait stitched from its
arm camera, with the upper part of Mount Sharp on the horizon.
NASA/JPL-Caltech/MSSS, [PIA19142](https://photojournal.jpl.nasa.gov/catalog/PIA19142).*

**2. Mars's air is thin, and every winter part of it freezes.** The median reading is
835 Pa, 0.8% of the air pressure at sea level on Earth (101,325 Pa).
That air is almost all carbon dioxide (CO2). Over the pole that is in winter it gets cold enough
for CO2 to freeze out of the air as frost on the polar cap; in spring the frost turns back into
gas. So the pressure over the whole planet rises and falls with the seasons: in Mars year 32
Curiosity read from 732 to 925 Pa, a swing of 21%.

<img src="https://raw.githubusercontent.com/AI4EPS/EPS88_PyEarth/main/data/images/T13/north_polar_cap_retreat.jpg" width="620">

*Mars's north polar cap seen by the Hubble Space Telescope, from early spring (October 1996) to
the start of summer (March 1997): the CO2 frost turns back into air and the cap shrinks. P. James, T.
Clancy, S. Lee and NASA/STScI, via [Wikimedia Commons](https://commons.wikimedia.org/wiki/File:Seasonal_Changes_In_Mars%27_North_Polar_Ice_Cap_(opo9715b).jpg).*

**3. Why it matters: Mars did lose most of its air, but slowly.** Early Mars was warmer than today.
NASA's MAVEN orbiter measured gas escaping from the top of Mars's air to
space at 2 to 3 kg every second, mostly hydrogen and oxygen atoms split from water, not the carbon
dioxide of the air. Run back over Mars's history, escape removed at least about 0.8 bar of CO2
(96 times today's air), and it is the main reason Mars became cold and dry
(Jakosky and others, 2018, [doi:10.1016/j.icarus.2018.05.030](https://doi.org/10.1016/j.icarus.2018.05.030)).
Today's air weighs about 32 million billion kg (its pressure times the area
of Mars, divided by gravity). Even if all 3 kg per second came out of that air, escape would remove
only 36 parts in a billion of it over Curiosity's 12 years: far
too little to see. So the 6% drop is something else.

**4. Three things move the barometer at once, and a model pulls them apart.** The season moves it,
through the polar caps. The rover's height moves it: Curiosity has been driving up Mount Sharp, a
mountain about 5 km high in the middle of the crater, and climbed 848 m in these years;
the higher you stand, the less air is above you. And the weather moves it, above all dust storms.
You cannot tell the three apart by looking at the readings. A model with one part for each can
measure each one, and you can test it by forecasting a year it has never seen.

body. NASA publishes one pressure value for each **sol** (a Mars day, 24 hours 40 minutes) on its
public weather page, and the rover's height from its map of the rover's drives. Your training file
holds the 4,110 sols with a pressure from 2012-08-16 to 2024-11-11, Mars years 31 to 37.
**One row is one sol:** the date on Earth, the season, the Mars year, the pressure, the day's lowest

**6. What earlier studies found.**

- **Haberle and others (2014)** read REMS's first 100 sols. The sensor worked well and showed the
  CO2 cycle, and daily pressure swings (the "thermal tides") several times larger than other
  landers had seen, partly because the crater amplifies them.
  [doi:10.1002/2013JE004488](https://doi.org/10.1002/2013JE004488)
- **Guzewich and others (2019)** describe the 2018 global dust storm from Curiosity: the dust hid
  97% of the Sun's ultraviolet light at the ground, the day-to-night temperature range fell by 30 K,
  and the pressure wave that rises and falls twice a day grew to an amplitude of 40 Pa.
  [doi:10.1029/2018GL080839](https://doi.org/10.1029/2018GL080839)
- **Lange and others (2022)** compared the pressure measured by two landers 40 years apart, Viking 1
  and InSight: the CO2 cycle changed by no more than 8 Pa. They saw hints of a larger north
  polar cap after the 2018 storm, with no agreed cause.
  [doi:10.1029/2022JE007190](https://doi.org/10.1029/2022JE007190)
- **de la Torre Juárez and others (2024)** compared the Gale record before and after the 2018
  storm and found the pressure lower than usual after it, a deficit later dust storms did not erase, and,
  from an orbiter, a larger north polar cap. [doi:10.1029/2023JE007810](https://doi.org/10.1029/2023JE007810)

## 1. How does the air pressure change through a Mars year, and from year to year?

**The season is counted by `ls`, Mars's position around the Sun, in degrees.** Ls = 0 is the start
of northern spring, 90 of northern summer, 180 of northern autumn and 270 of northern winter. One
full circle, a Mars year, takes 668 sols; Mars year 32 began on
2013-08-01 on Earth. Curiosity sits in Gale crater, a little south of the equator.

**Does the record sink from year to year?** The next cell draws every reading against time, and
marks the highest reading of each Mars year with its value.

```python
plt.figure(figsize=(10, 4.5))
plt.plot(train["sol"], train["pressure"], ".", markersize=2, color="0.5")
for year in range(31, 38):
    one_year = train[train["mars_year"] == year]
    top = one_year.iloc[np.argmax(one_year["pressure"])]     # the year's highest reading
    plt.plot(top["sol"], top["pressure"], "v", color="tab:red", markersize=8)
    plt.text(top["sol"], top["pressure"] + 8, f"{top['pressure']:.0f}", ha="center", color="tab:red")
    plt.text(np.mean(one_year["sol"]), 665, f"year {year}", ha="center", fontsize=9)
    plt.axvline(np.min(one_year["sol"]), color="0.8", lw=1)  # where each Mars year starts
plt.ylim(655, 950)
plt.xlabel("sol (Mars days since landing)")
plt.ylabel("pressure (Pa)")
plt.title(f"Air pressure at Curiosity, {len(train)} sols: each year's highest reading in red")
plt.show()
```

## 2. Can the seasons alone predict the pressure, or does the rover's climb matter?

**Mars's pressure needs more than the one wave a year that CO2 on Earth needed.** In the CO2 week
a pair of seasonal columns did the job:

**Seasonal column:** A column that rises and falls once a year, so the fitted curve can rise and fall with the seasons.

One pair, the sine and the cosine of the month's angle, was enough there, because CO2 rises and
falls once a year. Mars's pressure rises and falls twice, with highs of different heights, and one
pair can draw only one high.

**More pairs draw more shapes.** A second pair goes round twice a year, a third three times. With
the angle of the season in radians, $\theta$ = `np.radians(ls)`, the pairs are

$$\sin\theta,\ \cos\theta,\quad \sin 2\theta,\ \cos 2\theta,\quad \sin 3\theta,\ \cos 3\theta,\ \dots$$

and linear regression finds how much of each to add:

$$p = a + b_1 \sin\theta + c_1 \cos\theta + b_2 \sin 2\theta + c_2 \cos 2\theta + \dots$$

**Linear regression:** Draw the best straight line. Best means the smallest root mean squared error: the line that misses its points least.

With many columns it finds the best mix of all of them in the same way. The next cell draws the sine of each of the first three pairs.

```python
ls_values = np.arange(0, 361)
plt.figure(figsize=(9, 3.8))
for n in [1, 2, 3]:
    plt.plot(ls_values, np.sin(np.radians(n * ls_values)), lw=2, label=f"pair {n}: the sine of {n} × the angle")
plt.axhline(0, color="k", lw=0.8)
plt.xticks([0, 90, 180, 270, 360])
plt.xlabel("season, Ls (degrees)")
plt.ylabel("value of the column")
plt.title("The building blocks: waves that go round 1, 2 and 3 times a Mars year")
plt.legend(loc="upper center", bbox_to_anchor=(0.5, -0.22), ncol=3)
plt.show()
```

presses less: the year-to-year fall is the climb, not the planet.

**Air pressure falls with height in a known way.** Each metre up takes away the weight of one
metre of air, and the thinner the air, the less a metre of it weighs. So the pressure falls by the
same *fraction* with every metre:

$$p = p_0\, e^{-z/H} \qquad\text{so}\qquad \ln p = \ln p_0 - \frac{z}{H}$$

- $p$ is the pressure and $z$ the height, `elev_m`;
- $p_0$ is the pressure at height 0, which changes with the season;
- $H$ is the **scale height**: the climb over which the pressure falls by a factor of e, about 2.7.

In logs, the height is one more straight-line column. So fit `np.log(pressure)` on the seasonal
columns and `elev_m`: the seasonal columns give $\ln p_0$, the coefficient of `elev_m` is
$-1/H$, and `np.exp` turns a prediction back into pascals.

## 3. How well does the model forecast a Mars year it has never seen?

**A model is judged on data it did not learn from.**

**A fair baseline for a new year** is not the average of all sols, which ignores the season, but
