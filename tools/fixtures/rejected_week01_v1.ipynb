## Values, variables, and sentences

A **variable** is a name for a value. `=` does not mean "equals" — it means *put this value
under this name*, and the name can then be used anywhere below.

Values come in kinds, and `type()` tells you which: a whole number is an `int`, a number with a
decimal point is a `float`, text in quotes is a `str`, and `True` / `False` is a `bool`. Text
and numbers behave differently: `+` between two numbers adds them, and `+` between two strings
glues them end to end.

The comfortable way to print a sentence with a number in it is an **f-string**: put `f` in front
of the quotes and any name in `{}` braces gets replaced by its value. Writing `{value:.1f}`
instead rounds to one digit after the point.

`>` compares two numbers and gives you back `True` or `False` — that is where a `bool` comes
from. **Coming later:** comparisons are really week 2's topic, and week 2 uses them to make
decisions. Today they appear only here and inside the self-checks; no answer cell asks you to
write one.

## Finding the birthquake

The birthquake is the **largest** magnitude of the day, so `max(mags)` gives us the number. But
a number on its own is not an answer — we want to know *where* it was.

That is what `.index()` is for. `mags.index(6.5)` gives back the **position** of the value 6.5
in `mags`, and because the six lists line up, that same position read out of `places` and
`depths` describes the same earthquake. The dot in `mags.index(...)` is new: it means *the
`index` operation belonging to this particular list*.

```python
biggest = max(mags)
i = mags.index(biggest)

print(f"The birthquake was magnitude {biggest}, {places[i]}.")
print(f"It happened at {times[i]}, {depths[i]} km below the surface.")
print(f"The smallest earthquake in the file was magnitude {min(mags)}.")
```

## A map is a scatter plot

A map needs no special library. Longitude is just a number that runs from −180 to 180, latitude
runs from −90 to 90, so `plt.scatter(lons, lats)` puts every earthquake in its place.

Dots in a blank rectangle are unreadable, so we draw the world's coastline underneath first.
`coast` was loaded in the setup cell and holds nothing but longitude and latitude columns; the
file has a blank row between one coastline segment and the next, which makes `plt.plot` lift the
pen instead of ruling a line across the ocean. That is why the whole world draws in one line of
code.

`set_aspect("equal")` keeps one degree of longitude the same size on the page as one degree of
latitude, so the shapes are not stretched.

```python
plt.plot(coast.lon, coast.lat, color="0.6", lw=0.6)
plt.scatter(lons, lats, s=30)
plt.xlabel("longitude (degrees east)")
plt.ylabel("latitude (degrees north)")
plt.title(f"M4.5+ earthquakes on 2003-12-22 (n = {len(mags)})")
plt.gca().set_aspect("equal")
plt.xlim(-180, 180)
plt.ylim(-90, 90)
plt.show()
```

## What you are looking at

Seventeen dots were noise. Several thousand of the same dots are a planet.

Almost all of them fall in **narrow belts** — a ring around the Pacific, a line straight down the
middle of the Atlantic, a band running from the Mediterranean through Iran and the Himalaya. The
belts are narrow: a few hundred kilometres wide against ten thousand kilometres of quiet
continent on either side. And the empty ground is not simply somewhere nobody was listening —
look at the ridge running south-west from Africa, or the line below New Zealand. Those are the
emptiest, furthest-from-anywhere places on the planet, and the catalogue has earthquakes there.

Those belts are the **edges of the tectonic plates**. That is not something this map proves on
its own — it is the interpretation that seismologists arrived at in the 1960s, when maps very
like this one were first drawn and the narrow belts turned out to line up with where the sea
floor is made, destroyed and torn sideways (Isacks, Oliver & Sykes, *Seismology and the new
global tectonics*, 1968). What the map does show, on its own, is that the belts are there, that
they are narrow, and that they join into a few continuous systems rather than stopping at
random.

Look closely and a few dots do sit well inside a plate. Most earthquakes are on a boundary; not
all of them are.

## Roughly ten to one

Three numbers, all near ten. Step up one unit of magnitude and there are about ten times fewer
earthquakes; step up two units and about a hundred times fewer.

This is an old and famous result — Gutenberg and Richter published it for California in 1944
(*Frequency of earthquakes in California*), and it is why seismologists can say anything at all
about the size of earthquake nobody has lived through. It is an **empirical rule**, not a law of
arithmetic: "about ten" is the convention, our year gives numbers near it, and a different year
or a different region would give somewhat different ones. Our own catalogue is also mixing
several different ways of measuring magnitude, which is one more reason to read these as rough.

## The other end of the scale

The rule cuts both ways: it also says there should be about ten times *more* M3.5s than M4.5s,
and a hundred times more M2.5s. We cannot check that here, because we asked the catalogue for
magnitude 4.5 and up, so the smaller ones are simply not in our file.

And if we did ask for them, we would not get all of them. This is the first thing to know about
any catalogue, and it comes up again and again this term:

> **A catalogue lists what somebody's instruments recorded, not what happened. Where there are
> no seismometers there are no earthquakes in the file.**

A magnitude 6 is recorded by instruments on the far side of the planet. A magnitude 1 is
recorded by a seismometer a few kilometres away, or by nothing at all. So the small end of the
catalogue is not a measurement of the Earth — it is a measurement of where people have put
instruments. Your homework starts exactly there.
