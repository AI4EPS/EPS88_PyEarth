### 2. Why it matters: nobody can visit a planet's core

**Earth's core lies under the whole rocky mantle, out of reach; Psyche may show us a core.**

### 3. An asteroid is read from its colour

**An asteroid is a point of light, so astronomers read its make-up from the colours of that light.**
Each mineral absorbs some colours of sunlight more than others. The rock minerals olivine and
pyroxene absorb near 1,000 nm, in the near infrared just beyond what our eyes see (Reddy and
others, 2015, [arXiv:1502.05008](https://arxiv.org/abs/1502.05008)). So astronomers measure an asteroid's **reflectance spectrum**: the light it
reflects at each wavelength, divided by the light it reflects in green (550 nm). They sort spectra
into classes by shape, and group the classes into three **complexes**. This notebook calls each
complex a **kind**:

- **S** (stony): rises steeply toward the red and dips near 1,000 nm, the dip of olivine and pyroxene.
  Gaia's colours (step 5) stop at about 1,000 nm, so they see only the start of this dip, where an S
  spectrum begins to fall after about 800 nm;
- **C** (carbon-rich): nearly flat;
- **X**: rises gently with no dip. Psyche is an X, but so are asteroids that only share its colour.

- **Albedo**, the share of sunlight a surface reflects (0.1 means a tenth), comes from infrared
  telescopes. It is measured head-on, with the Sun behind the observer, and compared with a plain
  white disc; a very bright surface can reflect as much as that disc or more, so a few rows of the
  file reach 1. NASA's **WISE** measured the heat glow of more than 100,000 main-belt asteroids,
  which gives each one's size; size and visible brightness together give the albedo (Masiero and
  others, 2011).
- **Colour** comes from ESA's **Gaia** spacecraft, built to map the stars of the Milky Way, which
  also watches asteroids. Its third data release gives reflectance spectra of 60,518 asteroids in
  16 bands from 374 to 1,034 nm, observed from 2014 to 2017: each is the asteroid's light divided
  by the light of stars like the Sun (Galluccio and others, 2023).
- **Classes** are letters from two systems built from telescope observations: Tholen's (1984),
  from colours and albedo, and SMASSII (Bus and Binzel, 2002), from spectra.

### 6. What earlier research found

**For more than forty years astronomers have found that the kinds are not spread evenly through
the belt.**

- **Gradie and Tedesco (1982)** sorted asteroids into colour classes and found that the share of
  each class changes steadily with distance from the Sun, from 1.8 to 5.2 AU. They read this as
  asteroids having formed near where they orbit now.
  [doi:10.1126/science.216.4553.1405](https://doi.org/10.1126/science.216.4553.1405)
- **Masiero and others (2011)** measured the size and albedo of more than 100,000 main-belt
  asteroids with WISE. Albedo has two humps, dark and bright, and the average albedo changes from
  the inner to the outer belt.
  [doi:10.1088/0004-637X/741/2/68](https://doi.org/10.1088/0004-637X/741/2/68)
- **DeMeo and Carry (2014)** mapped the belt with many smaller asteroids and found it more mixed,
  by size and distance, than had been known: a sign that the belt was stirred, for example as the
  planets moved. [doi:10.1038/nature12908](https://doi.org/10.1038/nature12908)
- **Delbo and others (2026)** trained a classifier on Gaia spectra of asteroids with known classes
  and named a class for every asteroid in the Gaia release. S was sorted well; X mixed with its
  neighbours, as Gaia's short range of wavelengths leads one to expect.
  [doi:10.3389/fspas.2026.1774478](https://doi.org/10.3389/fspas.2026.1774478)

**What the columns hold.**

- `a`, `e`, `i`: the orbit. `a` is the average distance from the Sun in **AU** (the main belt
  lies between about 2.1 and 3.3 AU); `e` says how stretched the orbit is (0 is a circle); `i` is
  its tilt in degrees.
- `diameter` in km, and `albedo`.
- `R374` to `R1034`: Gaia's reflectance in 16 bands, named by wavelength in nm, each **divided by
  the reflectance at 550 nm (green)**. So `R550` is always 1, a value above 1 means "redder than
  green", and a blank is a band Gaia flagged as unreliable.
- `spec_T` and `spec_B`: the class letters of the Tholen and SMASSII (Bus) systems. Most
  asteroids have neither.

**The next cell turns the class letters into the three kinds S, C and X** of step 3 at the top.
It reads the Bus letter first and the Tholen letter if Bus has none, the leaderboard's rule. A
letter that names none of the three (D, T, an uncertain class with a colon) gives no kind.
Tholen's dark P class counts as X here, as the leaderboard counts it. That is a choice: the newer
classification of Mahlke and others (2022, step 7) puts P with C instead. In the data, 13 of the 194 X asteroids are X only through Tholen's P. Then the cell keeps the
asteroids with a kind **and** all 14 bands from 418 to 990 nm, because every leaderboard asteroid
has those.

```python
# the letters of each kind, in each of the two systems
BUS = {"S": ["S", "Sq", "Sl", "Sa", "Sk", "Sr", "Q", "A", "K", "L", "Ld", "R", "V", "O"],
       "C": ["C", "Ch", "Cb", "Cg", "Cgh", "B"],
       "X": ["X", "Xc", "Xk", "Xe"]}
THOLEN = {"S": ["S", "A", "Q", "R", "V"],
          "C": ["C", "B", "G", "F"],
          "X": ["X", "M", "E", "P"]}


def kind_of(letter, system):
    """S, C or X for one class letter of one system, or None if it names none of them."""
    for kind, letters in system.items():
        if letter in letters:
            return kind
    return None


# Bus first; Tholen only where Bus gives no kind
kinds = []
for bus, tholen in zip(asteroids["spec_B"], asteroids["spec_T"]):
    kind = kind_of(bus, BUS)
    if kind is None:
        kind = kind_of(tholen, THOLEN)
    kinds.append(kind)
asteroids["complex"] = kinds

# the 14 colour bands every leaderboard asteroid has (550 nm is always 1, but kept for simplicity)
bands = ["R418", "R462", "R506", "R550", "R594", "R638", "R682", "R726", "R770", "R814",
         "R858", "R902", "R946", "R990"]
labelled = asteroids[asteroids["complex"].notna() & asteroids[bands].notna().all(axis=1)]
labelled = labelled.reset_index(drop=True)

print("asteroids with a kind:", asteroids["complex"].notna().sum())
print("... and all 14 colour bands:", len(labelled))
print(labelled["complex"].value_counts())
```

### Scoring a guess with three answers

**The F1 score you met in the explosion week works on one kind at a time.** There it asked two
questions of one label, "explosion":

**Precision / recall:** Of the ones you flagged, how many were right? Of the real ones, how many did you catch?

F1 is the balance of the two. Here there are three kinds. We score each kind the same way, that kind
against the other two, and average the three F1 scores. The average is the **macro F1**. The
leaderboard ranks you by it.

**Why the average and not the share right?** The kinds are not equally common. A rule that never
says X can still be right most of the time, and an average of the three F1 scores gives such a
rule 0 for X. Macro F1 runs from 0 to 1, and 1 means every asteroid named correctly.

**A simple case first:** six made-up asteroids, their true kinds and a guess. The next cell scores
one kind, S, with `f1_score` from the explosion week: a True for every asteroid that is S, against
a True for every asteroid the guess calls S.

```python
# six made-up asteroids: their true kind and a guess for each
toy_truth = np.array(["S", "S", "S", "C", "C", "X"])
toy_guess = np.array(["S", "S", "C", "C", "S", "S"])

# the F1 of one kind, S: that kind against the other two
toy_f1_S = f1_score(toy_truth == "S", toy_guess == "S")
print("F1 for S:", round(toy_f1_S, 3))
```

```python
# the main belt in three zones, split at the Kirkwood gaps at 2.5 and 2.82 AU
zone_edges = [2.1, 2.5, 2.82, 3.3]
zone_names = ["inner, 2.1-2.5 AU", "middle, 2.5-2.82 AU", "outer, 2.82-3.3 AU"]
in_belt = labelled[(labelled["a"] > 2.1) & (labelled["a"] < 3.3)].copy()
in_belt["zone"] = pd.cut(in_belt["a"], bins=zone_edges, labels=zone_names)

# the share of each kind in each zone
share_by_zone = pd.crosstab(in_belt["zone"], in_belt["complex"], normalize="index")
print(share_by_zone.round(2))

# three bars per zone, one per kind, each labelled with its share
zone_sizes = in_belt["zone"].value_counts().reindex(zone_names)
plt.figure(figsize=(8, 4.5))
for k, kind in enumerate(["S", "C", "X"]):
    bars = plt.bar(np.arange(3) + (k - 1) * 0.27, 100 * share_by_zone[kind], width=0.27,
                   color=KIND_COLOURS[kind], label=kind)
    plt.bar_label(bars, fmt="%.0f%%")
plt.xticks(range(3), [f"{name}\n{size} asteroids" for name, size in zip(zone_names, zone_sizes)])
plt.xlabel("zone of the main belt, nearest the Sun on the left")
plt.ylabel("share of the zone's classified asteroids (%)")
plt.ylim(0, 75)
plt.title(f"Stony S falls and carbon-rich C rises outward ({len(in_belt)} asteroids)")
plt.legend(title="kind")
plt.show()
```

## 4. Does a different model, or more information, sort them better?

**The SVM draws boundaries in a different way from a forest.**

**SVM:** Of all the lines separating the two groups, take the one leaving the widest gap.

A forest cuts each column at a value, so its boundaries are steps; an SVM (with its default kernel)
draws smooth curved ones. Because it measures distances across the columns, an albedo of 0.1 and
an orbit of 2.8 AU must first be put on the same footing: `StandardScaler` rescales each column
to an average of 0 and a spread of 1, and `make_pipeline(StandardScaler(), SVC())` joins the two,
as in the basalt week.
